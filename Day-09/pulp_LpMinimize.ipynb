```python
from pulp import *
model = LpProblem('LoanMinimizeProblem',LpMinimize)
#create var
emi_bank_a=2076
emi_bank_b=3227
emi_bank_c=4707
total_loan_required=10 #lakh

x1=LpVariable('Loan_from_bank_a',0,None,LpContinuous)
x2=LpVariable('Loan_from_bank_b',0,None,LpContinuous)
x3=LpVariable('Loan_from_bank_c',0,None,LpContinuous)

model += emi_bank_a*x1 + emi_bank_b*x2 + emi_bank_c*x3 , "Total_monthly_emi"

#total loan must be 10 lakhs
model+=x1+x2+x3==total_loan_required,"Total_loan_constraints"
model+= x1>=2,"Min_loan_bank_a"
model+= x3<=4,"Max_loan_bank_c"
model.solve()

for v in model.variables():
    print(v.name,"=",v.varValue)
print("Mimimum total monthly emi : Rs.",value(model.objective))
```
