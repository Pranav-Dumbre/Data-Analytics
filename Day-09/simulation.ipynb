```python
# monte carlo simulation
import numpy as np

# part1
def roll_dice():
    # simulation of rolling dice twice, minimum number we get is 1, max is 6
    # Add the results of the two simulation, so possible values be:
    #(1,1) or (1,2)...(6,6) .. sum will be between 2 and 12
    return np.sum(np.random.randint(1,7,2)) # Generate an  array of 2 random 
    #integer bet 1 and 6 (both inclusive) and add them

print(roll_dice())
```

```python
# part 2: 
def monte_carlo_simulation(runs=1000):
    results = np.zeros(2)
    for _ in range(runs):
        if roll_dice()==7:
            results[0] +=1
        else:
            results[1] +=1
    return results
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())
```

```python
results = np.zeros(1000)

for i in range(1000):
    results[i] = monte_carlo_simulation()[0]

print(results)
net_amount=(results * 5) -((1000-results)*1)
print(net_amount)

```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.hist(results, bins =15, edgecolor='black')
plt.show()
```

```python
# our wins/ loss
print(results.mean())
print(results.mean()*5)
print(results.mean()*4.75)
print(1000-results.mean())
print(results.mean()/1000)
```

```python
# monte carlo for investment returns
import numpy as np
import matplotlib.pyplot as plt

# pareameter
initial_investment =1000
mean_return = 0.15
std_dev = 0.15
num_simulation =1000
time_horizon = 10

# monte carlo simulation
simulated_end_values = []
for _ in range(num_simulation):
    investment_value = initial_investment
    for _ in range(time_horizon):
        random_return = np.random.normal(mean_return,std_dev)
        investment_value *=(1+ random_return)
    simulated_end_values.append(investment_value)

# results
mean_end_value = np.mean(simulated_end_values)
risk_of_loss = np.mean(np.array(simulated_end_values)< initial_investment)
print(f'Mean and value: ${mean_end_value:.2f}')
print(f'Probability of loss:{risk_of_loss *100:.2f}%')

# plotting the distribution
plt.hist(simulated_end_values, bins =30, edgecolor='k')
plt.title('Simulated Portfolio End values after 5 years')
plt.xlabel('Portfolio value ($)')
plt.ylabel('Frequency')
plt.show()
```
