```python
import pandas as pd
from scipy import stats
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
our_column = df['tip']
our_column = our_column.sample(n=30,replace=True)
meu = 3  # we assume that average tip is $3 and we want test it
t_statstics, p_value = stats.ttest_1samp(our_column,meu)
print('p-value:',p_value)

alpha = 0.5
if p_value < alpha:
    print('reject H0')
else:
    print('Do not reject H0')
```

```python
# 2 sample T testing
male_tips=df[df['sex']=='Male']['tip']
female_tips=df[df['sex']=='Female']['tip']

male_tips=male_tips.sample(n=30,replace=False)
female_tips=female_tips.sample(n=30,replace=False)

t_statstics, p_value = stats.ttest_ind(male_tips,female_tips)
print('p-value:',p_value)
print("t_statstics :",t_statstics)
alpha = 0.5
if p_value < alpha:
    print('reject H0')
else:
    print('Do not reject H0')
```

```python
# paired test for t.1 and t.2
# H0:Treatment is not effective
# H1/Ha: It is effective
df = pd.read_csv('hypothermia.csv')
df = df[['t.1','t.2']].dropna()
sample_df = df.sample(n=30, replace=False)

t_statistic, p_value = stats.ttest_rel(sample_df['t.1'],sample_df['t.2'])
print('P-value:',p_value)

if p_value < alpha:
    print('reject H0')
else:
    print('Do not reject H0')
```

```python
#ANOVA
df = sns.load_dataset('tips')
df
```

```python

df.rename(columns={'size':'cust_count'},inplace=True)
df['tip_to_total']=df['tip']/df['total_bill']

s1=df.tip_to_total[df.cust_count==1]
s2=df.tip_to_total[df.cust_count==2]
s3=df.tip_to_total[df.cust_count==3]
s4=df.tip_to_total[df.cust_count==4]
s5=df.tip_to_total[df.cust_count==5]
s6=df.tip_to_total[df.cust_count==6]

#ANOVA
fstat,pval=stats.f_oneway(s1,s2,s3,s4,s5,s6)
alpha=0.05
print(pval)
if pval < alpha:
    print('Reject H0')
else:
    print("Do not reject H0")
```
