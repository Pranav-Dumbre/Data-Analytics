```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('weight-height.csv')
df.head()
```

```python
df_male= df[df['Gender']=='Male']
df_female= df[df['Gender']=='Female']
```

```python
corr_overall = df[['Height','Weight']].corr()
corr_male = df_male[['Height','Weight']].corr()
corr_female = df_female[['Height','Weight']].corr()
```

```python
cov_overall = df[['Height','Weight']].cov()
cov_male = df_male[['Height','Weight']].cov()
cov_female = df_female[['Height','Weight']].cov()
```

```python
print('Correlation of overall\n',corr_overall,'\nCorrelation of male\n',corr_male,'\nCorrelation of female\n',corr_female,'\ncovar of overall\n',cov_overall,'\ncovar of male\n',cov_male,'\ncovar of female\n',cov_female)
```

```python
# scatter plot with corr value
fig, axes = plt.subplots(1,3,figsize=(18,6))
axes[0].scatter(df['Height'],df['Weight'],alpha=0.5,color='#453587')
axes[0].set_title(f'Overall corr:{corr_overall.loc['Height','Weight']:.2f}')
axes[0].set_xlabel('Height')
axes[0].set_ylabel('Weight')
axes[0].grid(True)

# Male data plot
axes[1].scatter(df_male['Height'],df_male['Weight'],alpha=0.5,color='#426C8C')
axes[1].set_title(f'Overall corr:{corr_male.loc['Height','Weight']:.2f}')
axes[1].set_xlabel('Height')
axes[1].set_ylabel('Weight')
axes[1].grid(True)

# Female data plot
axes[2].scatter(df_female['Height'],df_female['Weight'],alpha=0.5,color='#873561')
axes[2].set_title(f'Overall corr:{corr_female.loc['Height','Weight']:.2f}')
axes[2].set_xlabel('Height')
axes[2].set_ylabel('Weight')
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
# Heatmap
fig,[ax1,ax2,ax3] = plt.subplots(1,3,figsize=(18,6))
sns.heatmap(corr_overall,annot=True,fmt='.2f',cmap='coolwarm',ax=ax1)
ax1.set_title('Overall relation heatmap')
plt.xticks(rotation=45)
plt.yticks(rotation=0)

sns.heatmap(corr_male,annot=True,fmt='.2f',cmap='coolwarm',ax=ax2)
ax2.set_title('male relation heatmap')

sns.heatmap(corr_female,annot=True,fmt='.2f',cmap='coolwarm',ax=ax3)
ax3.set_title('female relation heatmap')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
