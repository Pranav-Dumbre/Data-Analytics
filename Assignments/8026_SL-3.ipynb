```python
#1. Load the tips dataset into a DataFrame.
import numpy as np
import pandas as pd
import seaborn as sns
df=sns.load_dataset('tips')
df
```

```python
# 2. Display the first 5 rows of the tips dataset.
df.head()
```

```python
# 3. Show all column names in the tips dataset.
df.columns
```

```python
# 4. Find the average total_bill.
df['total_bill'].mean()
```

```python
# 5. Find the maximum tip given.
df['tip'].max()
```

```python
# 6. Filter all rows where total_bill is greater than 20.
df[df.total_bill>20]
```

```python
# 7. Filter all rows where smoker is Yes.
df[df.smoker=='Yes']
```

```python
# 8. Show only total_bill, tip, and day columns.
df[['total_bill','tip','day']]
```

```python
# 9. Find the average tip grouped by day.
df.groupby('day',observed=True)['tip'].mean()
```

```python
# 10. Find the average total_bill grouped by time.
df.groupby('time',observed=True)['total_bill'].mean()
```

```python
# 11. Count how many records belong to each day.
df.groupby('day',observed=True)['day'].count()
```

```python
# 12. Count how many smokers and non-smokers are present.
smoker=df.groupby('smoker',observed=True)['smoker'].count()
smoker
```

```python
# 13. Find the total tip collected on each day.
df.groupby('day',observed=True)['tip'].sum()
```

```python
# 14. Find the average tip for lunch and dinner separately.
df.groupby('time',observed=True)['tip'].mean()
```

```python
# 15. Create a new column tip_percent = (tip / total_bill) * 100.
df['tip_percent']=(df.tip/df.total_bill)*100
df
```

```python
# 16. Find the row with the highest tip_percent.
df['tip_percent'].max()
```

```python
# 17. Find the average size of dining group by day.
df.groupby('day',observed=True)['size'].mean()
```

```python
# 18. Sort the tips dataset by tip in descending order.
df.sort_values(by='tip',ascending=False)
```

```python
# 19. Filter records where size is greater than or equal to 4.
df.rename(columns={'size':'cust_count'},inplace=True)
df[df.cust_count>=4]
```

```python
# 20. Find the average tip_percent for smokers vs non-smokers.
df.groupby('smoker',observed=True)['tip_percent'].mean()
```
