```python
# 14. Adj close and SMA 30 and SMA 40
df['SMA30'] = df['Adj Close'].rolling(30).mean()
df['SMA40'] = df['Adj Close'].rolling(40).mean()

plt.Figure()
plt.plot(df['Adj Close'] , label='Adj Close')
plt.plot(df.SMA30, label='SMA 30')
plt.plot(df.SMA40, label='SMA 40')
plt.xlabel('Days')
plt.ylabel('Price')
plt.legend()
plt.title('Adjusted close and SMA')
plt.show()
```

```python
# 15. SMA + EMA
df['SMA30'] = df['Adj Close'].rolling(30).mean()
df['EMA30'] = df['Adj Close'].ewm(span=30).mean()

plt.Figure()
plt.plot(df['Adj Close'], label='Adj Close')
plt.plot(df.SMA30, label='SMA 30')
plt.plot(df.EMA30, label='EMA 30')
plt.xlabel('Days')
plt.ylabel('Price')
plt.legend()
plt.title('SMA vs EMA')
plt.show()
```

```python
# 16. Scatter volume vs price b/w two dates
plt.Figure()
plt.scatter(df.Volume, df['Adj Close'])
plt.xlabel('Volume')
plt.ylabel('Adj Close')
plt.title('Volume vs close')
plt.show()
```

```python
# 17. Daily returns
df['Returns'] = df['Adj Close'].pct_change()

plt.Figure()
plt.plot(df.Returns)
plt.title('Daily Returns')
plt.show()
```

```python
# 18. volatility
df['Volatility'] = df.Returns.rolling(30).std()

plt.Figure()
plt.plot(df.Volatility)
plt.title('30 Days Volatility')
plt.show()
```

```python
# 19. Histogram
plt.Figure()
plt.hist(df.Returns.dropna(), bins=50, edgecolor='cyan', density=True)
plt.title('Returns Distribution')
plt.show()
```
