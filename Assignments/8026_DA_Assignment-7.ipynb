```python
# 1. Write a Pandas program to detect missing values of a given DataFrame.
# Display True or False.
import pandas as pd
import numpy as np

# Create the DataFrame
data = {
    'ord_no': [70001.0, np.nan, 70002.0, 70004.0, np.nan, 70005.0, np.nan, 70010.0, 70003.0, 70012.0, np.nan, 70013.0],
    'purch_amt': [150.50, 270.65, 65.26, 110.50, 948.50, 2400.60, 5760.00, 1983.43, 2480.40, 250.45, 75.29, 3045.60],
    'ord_date': ['2012-10-05', '2012-09-10', np.nan, '2012-08-17', '2012-09-10', '2012-07-27', '2012-09-10',
                 '2012-10-10', '2012-10-10', '2012-06-27', '2012-08-17', '2012-04-25'],
    'customer_id': [3002, 3001, 3001, 3003, 3002, 3001, 3001, 3004, 3003, 3002, 3001, 3001],
    'salesman_id': [5002.0, 5003.0, 5001.0, np.nan, 5002.0, 5001.0, 5001.0, np.nan, 5003.0, 5002.0, 5003.0, np.nan]
}

df = pd.DataFrame(data)

# Detect missing values
missing_values = df.isnull()

print(missing_values)
```

```python
# 2. Write a Pandas program to identify the column(s) of a given DataFrame which
# have at least one missing value.
missing_columns = df.isnull().any()
print(missing_columns)
```

```python
# 3. Write a Pandas program to count the number of missing values in each
# column of a given DataFrame.
count_missing = df.isnull().sum()
print(count_missing)
```

```python
# 4. Write a Pandas program to find and replace the missing values in a given
# DataFrame which do not have any valuable information.
fill1=df['ord_no'].fillna(0)
fill2=df['ord_date'].fillna('Unknown')
fill3=df['salesman_id'].fillna(-1)
print(fill1)
print(fill2)
print(fill3)
```

```python
# 5. Write a Pandas program to drop the rows where at least one element is
# missing in a given DataFrame.
df.dropna()
```

```python
# 6. Write a Pandas program to drop the columns where at least one element is
# missing in a given DataFrame.
df.dropna(axis=1)
```

```python
# 7. Write a Pandas program to drop the rows where all elements are missing in a given DataFrame.
df.dropna(how='all')
```

```python
# 8. Write a Pandas program to keep the rows with at least 2 NaN values in a given DataFrame.
df.isnull().sum(axis=1) >= 2
```

```python
# 9. Write a Pandas program to drop those rows from a given DataFrame in which
# specific columns have missing values
df.dropna(subset=['ord_no', 'salesman_id'])
```

```python
# 10. Write a Pandas program to keep the valid entries of a given DataFrame.
df.dropna()
```

```python
# 11. Write a Pandas program to calculate the total number of missing values in a DataFrame.
df.isnull().sum()
```

```python
# 12. Write a Pandas program to replace NaNs with a single constant value in
# specified columns in a DataFrame.
df.fillna({
    'ord_no': 0,            
    'ord_date': 'Unknown',
    'salesman_id': -1         
}, inplace=True)
df
```

```python
# 13. Write a Pandas program to replace NaNs with the value from the previous
# row or the next row in a given DataFrame.
df_ffill = df.fillna(method='ffill')
df_bfill = df.fillna(method='bfill')
print(df_ffill)
print(df_bfill)
```

```python
# 14. Write a Pandas program to replace NaNs with median or mean of the
# specified columns in a given DataFrame.

fill_mean= df['ord_no'].fillna(df['ord_no'].mean())
fill_median= df['salesman_id'].fillna(df['salesman_id'].median())
fill_unknown = df['ord_date'].fillna('Unknown')
print(fill_mean)
print(fill_median)
print(fill_unknown)
```

```python
# 15. Write a Pandas program to interpolate the missing values using the Linear
# Interpolation method in a given DataFrame.
# From Wikipedia, in mathematics, linear interpolation is a method of curve fitting
# using linear polynomials to construct new data points within the range of a
# discrete set of known data points.
# Apply linear interpolation to numeric columns
df_interpolated = df.interpolate(method='linear')

print(df_interpolated)
```

```python
# 16. Write a Pandas program to count the number of missing values of a specified
# column in a given DataFrame.
missing_ord_no = df['ord_no'].isnull().sum()
print("Missing values in 'ord_no':", missing_ord_no)

missing_salesman = df['salesman_id'].isnull().sum()
print("Missing values in 'salesman_id':", missing_salesman)
```

```python
# 17. Write a Pandas program to count the missing values in a given DataFrame.
df.isnull().sum().sum()
```

```python
# 18. Write a Pandas program to find the Indexes of missing values in a given DataFrame.
missing_indexes = df[df.isnull().any(axis=1)].index.tolist()
print("Indexes with missing values:", missing_indexes)
```

```python
# 19. Write a Pandas program to replace the missing values with the most frequent
# values present in each column of a given DataFrame.
for col in df.columns:
    df[col].fillna(df[col].mode()[0])

print(df)
```

```python
# 20. Write a Pandas program to create a hitmap for more information about the
# distribution of missing values in a given DataFrame.
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'ord_no': [70001.0, np.nan, 70002.0, 70004.0, np.nan, 70005.0, np.nan, 70010.0, 70003.0, 70012.0, np.nan, 70013.0],
    'purch_amt': [150.50, 270.65, 65.26, 110.50, 948.50, 2400.60, 5760.00, 1983.43, 2480.40, 250.45, 75.29, 3045.60],
    'ord_date': ['2012-10-05', '2012-09-10', np.nan, '2012-08-17', '2012-09-10', '2012-07-27', '2012-09-10',
                 '2012-10-10', '2012-10-10', '2012-06-27', '2012-08-17', '2012-04-25'],
    'customer_id': [3002, 3001, 3001, 3003, 3002, 3001, 3001, 3004, 3003, 3002, 3001, 3001],
    'salesman_id': [5002.0, 5003.0, 5001.0, np.nan, 5002.0, 5001.0, 5001.0, np.nan, 5003.0, 5002.0, 5003.0, np.nan]
}

df = pd.DataFrame(data)

plt.figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar=False, cmap='coolwarm', yticklabels=True)
plt.title("Heatmap of Missing Values in DataFrame")
plt.show()
```
