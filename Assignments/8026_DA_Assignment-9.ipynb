1. Write a Pandas program to create
a) Datetime object for Jan 15 2012.
b) Specific date and time of 9:20 pm.
c) Local date and time.
d) A date without time.
e) Current date.
f) Time from a datetime.
g) Current local time.

```python
# a) Datetime object for Jan 15 2012
date_a = pd.to_datetime('2012-01-15')
print(f"a) Specific Date: {date_a}")

# b) Specific date and time of 9:20 pm
# Note: 9:20 pm is 21:20 in 24-hour format
date_b = pd.to_datetime('2012-01-15 21:20:00')
print(f"b) Specific Time: {date_b}")

# c) Local date and time (using standard library via Pandas)
date_c = pd.Timestamp.now()
print(f"c) Local Date and Time: {date_c}")

# d) A date without time
date_d = pd.Timestamp.now().date()
print(f"d) Date without Time: {date_d}")

# e) Current date
date_e = pd.to_datetime('today')
print(f"e) Current Date: {date_e}")

# f) Time from a datetime
date_f = pd.Timestamp.now().time()
print(f"f) Time from Datetime: {date_f}")

# g) Current local time
# While .now() provides both, .time() extracts the specific clock time
date_g = pd.Timestamp.now().strftime('%H:%M:%S')
print(f"g) Current Local Time: {date_g}")
```

2. Write a Pandas program to create
a) a specific date using timestamp.
b) date and time using timestamp.
c) a time adds in the current local date using timestamp.
d) current date and time using timestamp.

```python
date_a = pd.Timestamp('2025-12-25')
print(f"a) Specific Date: {date_a}")

# b) Date and time using timestamp
# Passing both date and time components
date_b = pd.Timestamp(2026, 4, 10, 15, 30)
print(f"b) Date and Time: {date_b}")

# c) A time added to the current local date using timestamp
# We take the current date (floored) and add a Timedelta (offset)
current_date = pd.Timestamp.now().floor('D')
date_c = current_date + pd.Timedelta(hours=14, minutes=45)
print(f"c) Current Date with added time (2:45 PM): {date_c}")

# d) Current date and time using timestamp
date_d = pd.Timestamp.now()
print(f"d) Current Date and Time: {date_d}")
```

12. Write a Pandas program to convert year and day of year into a single
datetime column of a dataframe.

```python
df = pd.DataFrame({
    'year': [2024, 2025, 2026],
    'day_of_year': [32, 100, 156]
})
df['date'] = pd.to_datetime(df.year.astype(str), format='%Y') + \
             pd.to_timedelta(df.day_of_year -1, unit='D')

print(df)
```
