```python
# 50. Create two DataFrames with a common column and merge them.
data = {
    'Name': ['Dev', 'rudra','nayan','ashish','vijay'],
    'roll': [21,22,23,24,25],
}
records = {
    'roll': [21,22,23,24,25],
    'Marks': [100,90,90,76,86]
}

students=pd.DataFrame(data)
marks=pd.DataFrame(records)
merged_df = pd.merge(students, marks, on='roll')
merged_df
```
