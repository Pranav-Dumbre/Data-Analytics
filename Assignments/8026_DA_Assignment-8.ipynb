```python
#3. Write a Python program to draw a line using given axis values taken from a
# text file, with suitable label in the x axis, y axis and a title.

x_values = []
y_values = []

with open("test.txt", "r") as file:
    for line in file:
        parts = line.strip().split()
        if len(parts) == 2:  # Ensure there are two values per line
            x, y = parts
            x_values.append(float(x))
            y_values.append(float(y))

plt.plot(x_values, y_values, marker='o', linestyle='-', color='b')
plt.xlabel("X Axis Label")
plt.ylabel("Y Axis Label")
plt.title("Line Graph from File Data")
plt.show()
```

```python
# 4. Write a Python program to draw line charts of the financial data of Alphabet 
# Inc. between October 3, 2016 to October 7, 2016. 
# Sample Financial data (fdata.csv): 
# Date,Open,High,Low,Close 
# 10-03-16,774.25,776.065002,769.5,772.559998 
# 10-04-16,776.030029,778.710022,772.890015,776.429993 
# 10-05-16,779.309998,782.070007,775.650024,776.469971 
# 10-06-16,779,780.47998,775.539978,776.859985 
# 10-07-16,779.659973,779.659973,770.75,775.080017 
# The code snippet gives the output shown in the following screenshot: 
import io
csv_data = """Date,Open,High,Low,Close
10-03-16,774.25,776.065002,769.5,772.559998
10-04-16,776.030029,778.710022,772.890015,776.429993
10-05-16,779.309998,782.070007,775.650024,776.469971
10-06-16,779,780.47998,775.539978,776.859985
10-07-16,779.659973,779.659973,770.75,775.080017"""

df = pd.read_csv(io.StringIO(csv_data))
df['Date'] = pd.to_datetime(df['Date'], format='%m-%d-%y')

plt.figure(figsize=(8, 6))
plt.plot(df['Date'], df['Open'], label='Open')
plt.plot(df['Date'], df['High'], label='High')
plt.plot(df['Date'], df['Low'], label='Low')
plt.plot(df['Date'], df['Close'], label='Close')

plt.title('Financial Data of Alphabet Inc. (Oct 3 - Oct 7, 2016)')
plt.xlabel('Date')
plt.ylabel('Price')

plt.xticks(df['Date'])
plt.legend()

plt.show()
```

```python
# 14. Write a Python program to display the grid and draw line charts of the 
# closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. 
# Customized the grid lines with rendering with a larger grid (major grid) and a 
# smaller grid (minor grid).Turn on the grid but turn off ticks. 
# The code snippet gives the output shown in the following screenshot:
fig, ax = plt.subplots()
plt.grid(linestyle='-',linewidth=0.5)
x=np.arange(772.5,778,0.5)
plt.xlim(df['Date'].min(), df['Date'].max())
plt.ylim(772.5,777)
plt.plot(df['Date'], df['Close'], color='red', marker='o')
plt.xlabel('Date')
plt.ylabel('Price')
ax.minorticks_on()
ax.grid(which='major', linestyle='-', linewidth='0.5', color='red')
ax.grid(which='minor', linestyle=':', linewidth='0.5', color='black')
plt.xticks(df['Date'])
plt.yticks(x)
plt.show()
```

```python
# 15. Write a Python program to create multiple plots. 
# The code snippet gives the output shown in the following screenshot: 
fig = plt.figure(figsize=(8, 6))
ax1 = plt.subplot2grid((2, 3), (0, 0), colspan=3)
ax2 = plt.subplot2grid((2, 3), (1, 0))
ax3 = plt.subplot2grid((2, 3), (1, 1))
ax4 = plt.subplot2grid((2, 3), (1, 2))
for ax in [ax1, ax2, ax3, ax4]:
    ax.set_xticks([])
    ax.set_yticks([])
plt.tight_layout()
plt.show()
```
