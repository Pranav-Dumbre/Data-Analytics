```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= sns.load_dataset('titanic')
df
```

```python
df=df[['sex','survived']]
df.dropna(inplace=True)
```

```python
total = len(df)
print(f'Total passenger:{total}')
```

```python
# Marginal probabilty gender
total_female = len(df[df['sex']=='female'])
print(f'Total female:{total_female}')
total_male = len(df[df['sex']=='male'])
print(f'Total male:{total_male}')
p_female = total_female/total
p_male = total_male/total
print(f'Probabilty of female:{p_female:.3f}')
print(f'Probability male:{p_male:.3f}')
```

```python
# Marginal probability of survival
total_survival = len(df[df['survived']==1])
print(f'Total survived:{total_survival}')
total_did_not_survival = len(df[df['survived']==0])
print(f'Total not survived:{total_did_not_survival}')

p_survived = total_survival /total
p_not_survived = total_did_not_survival/total
print(f'p(survived=1):{p_survived:.3f}')
print(f'p(survived=0):{p_not_survived:.3f}')
```

```python
# joint probability
p_female_survived = len(df[(df['sex']=='female')&(df['survived']==1)])/total
p_female_not_survived = len(df[(df['sex']=='female')&(df['survived']==0)])/total
p_male_survived = len(df[(df['sex']=='male')&(df['survived']==1)])/total
p_male_not_survived = len(df[(df['sex']=='male')&(df['survived']==0)])/total

print('\nJoint probabilites:')
print(f'p(female,survived=1):{p_female_survived:.3f}')
print(f'p(female,survived=0):{p_female_not_survived:.3f}')
print(f'p(male,survived=1):{p_male_survived:.3f}')
print(f'p(male,survived=0):{p_male_not_survived:.3f}')
```

```python
print('\nConditional probabilities:')
# p(survive | female)=p(survive and female)/p(female)
p_survived_given_female = p_female_survived / p_female
print(f'p(survived=1 | sex=female):{p_survived_given_female:.3f}')
p_survived_given_male = p_male_survived / p_male
print(f'p(survived=1 | sex=male):{p_survived_given_male:.3f}')
```
